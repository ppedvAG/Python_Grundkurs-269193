Strings ersetzen

```python
data["YearlyChange"] = data["YearlyChange"].replace(" %", "", regex=True)
```

```python
def fixPct(col):
    data[col] = data[col].replace(" %", "", regex=True)
```

```python
fixPct("UrbanPop %")
```

```python
fixPct("WorldShare")
```

```python
data
```

```python
def fixNA(col):
    data[col] = data[col].replace("N.A.", 0, regex=True)
```

```python
fixNA("Med.Age")
```

```python
for col in data.columns:
    fixNA(col)
```

```python
data
```
